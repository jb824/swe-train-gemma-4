This notebook is a thin entrypoint for the file-configured CLI. Install the package in this kernel environment first. Edit `train.yaml`; runtime budgets come from the referenced harness. GPU support and full-context memory must be checked on your hardware.

```python
import subprocess, sys
from pathlib import Path
CONFIG = Path("train.yaml").resolve()
PROFILE = "rocm32"  # rocm32, nvidia48x2, or l4x4
def cli(*args):
    subprocess.run([sys.executable, "-m", "swegemma_training.cli", *map(str, args)], check=True)
if not CONFIG.exists():
    cli("config", "init", "--profile", PROFILE, "--output", CONFIG)
print(f"Edit {CONFIG} before running the remaining cells.")
```

```python
cli("config", "validate", "--config", CONFIG)
cli("doctor", "--config", CONFIG)
```

```python
cli("prepare", "--config", CONFIG)
```

```python
cli("train", "--config", CONFIG)
# To resume instead: cli("train", "--config", CONFIG, "--resume", "latest")
```
